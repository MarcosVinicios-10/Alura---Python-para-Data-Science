```python
import pandas as pd
vendas = pd.read_json('dados_vendas_clientes.json')
```

<p>O desafio do projeto 1: vendas online será abrir a base de dados com Pandas e aplicar o json_normalize.</p>

```python
vendas
```

```python
vendas = pd.json_normalize(vendas['dados_vendas'])
vendas
```

<P><ul>Remover os dados em listas dentro do DataFrame</ul>
<ul>Verificar os tipos de dados;</ul>
<ul>Identificar colunas numéricas;</ul>
<ul>Transformar a coluna numérica para o tipo numérico.</ul>
</P>

```python
vendas = vendas.explode(['Cliente',	'Valor da compra'])
vendas = vendas.reset_index(drop=True)
vendas
```

```python
vendas.info()
```

```python
vendas['Valor da compra'] = vendas['Valor da compra'].apply(lambda x: x.replace('R$','').replace(',','.').strip())
vendas
```

```python
vendas['Valor da compra'] = vendas['Valor da compra'].astype('float64')
vendas['Data de venda'] = pd.to_datetime(vendas['Data de venda'])
vendas.info()
```

```python
vendas
```

```python
vendas['Cliente'] = vendas['Cliente'].str.replace('[^a-zA-Zà-úÀ-ÚçÇ\s]','',regex=True)
vendas['Cliente'] = vendas['Cliente'].str.lower()
vendas
```

```python
vendas.groupby(vendas['Data de venda'].dt.strftime('%Y-%m'))['Valor da compra'].mean()
```
