```python
import pandas as pd
```

```python
dados_hospedagem = pd.read_json('dados_hospedagem.json')
moveis_disponiveis = pd.read_json('moveis_disponiveis.json')
```

```python
moveis_disponiveis
```

```python
dados_hospedagem = pd.json_normalize(dados_hospedagem['info_moveis'])
```

```python
dados_hospedagem 
```

<P>Desagrupando os dados</P>

```python
colunas = list(dados_hospedagem.columns)
colunas
```

```python
dados_hospedagem = dados_hospedagem.explode(colunas[3:])
dados_hospedagem =  dados_hospedagem.reset_index(drop=True)
```

```python
dados_hospedagem
```

```python
dados_hospedagem.info()
```

<P>Alterando o tipo das colunas</P>

```python
colunas_numericas = ['max_hospedes','quantidade_banheiros',	'quantidade_quartos','quantidade_camas']
dados_hospedagem[colunas_numericas] = dados_hospedagem[colunas_numericas].astype('int64')
```

```python
dados_hospedagem['avaliacao_geral'] = dados_hospedagem['avaliacao_geral'].astype('float64')
```

```python
dados_hospedagem.info()
```

```python
dados_hospedagem['preco'] = dados_hospedagem['preco'].apply(lambda x: x.replace('$','').replace(',','').strip())
dados_hospedagem[['taxa_deposito',	'taxa_limpeza']] = dados_hospedagem[['taxa_deposito','taxa_limpeza']].map(lambda x: x.replace('$','').replace(',','').strip())
dados_hospedagem
```

```python
dados_hospedagem[['taxa_deposito',	'taxa_limpeza','preco']] = dados_hospedagem[['taxa_deposito',	'taxa_limpeza','preco']].astype('float64')
```

```python
dados_hospedagem.info()
```

```python
dados_hospedagem['modelo_cama'].unique()
```

```python
dados_hospedagem['modelo_cama'] = dados_hospedagem['modelo_cama'].astype('category')
```

```python
dados_hospedagem.info()
```

Tokenização

Retirando as letras minúsculas 

```python
dados_hospedagem['descricao_local'] = dados_hospedagem['descricao_local'].str.lower()
dados_hospedagem
```

```python
dados_hospedagem['descricao_local'] = dados_hospedagem['descricao_local'].str.replace('[^a-zA-Z0-9\-\']',' ',regex=True)
dados_hospedagem['descricao_local'] = dados_hospedagem['descricao_local'].str.replace('(?<!\w)-(?!\w)',' ',regex=True)
dados_hospedagem
```

```python
dados_hospedagem['descricao_local'] = dados_hospedagem['descricao_local'].str.split()
dados_hospedagem
```

```python
dados_hospedagem['comodidades'] = dados_hospedagem['comodidades'].apply(lambda x: x.replace('{','').replace(',',' ').replace('}','').replace('"',''))
dados_hospedagem['comodidades'] = dados_hospedagem['comodidades'].str.split()
dados_hospedagem.head()
```

```python
dados_hospedagem['descricao_vizinhanca'] = dados_hospedagem['descricao_vizinhanca'].str.replace('[^a-zA-Z0-9\-\']',' ',regex=True)
dados_hospedagem['descricao_vizinhanca'] = dados_hospedagem['descricao_vizinhanca'].str.replace('(?<!\w)-(?!\w)',' ',regex=True)
dados_hospedagem['descricao_vizinhanca'] = dados_hospedagem['descricao_vizinhanca'].str.split()
dados_hospedagem
```
