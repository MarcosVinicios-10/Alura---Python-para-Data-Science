```python
import pandas as pd 
import numpy as np 
import json
from decimal import Decimal
dados = pd.read_json('dataset-telecon.json')
```

```python
with open('dataset-telecon.json',encoding='UTF-8') as f:
    json_original = json.load(f)
json_original
```

```python
dados = pd.json_normalize(json_original)
dados.info()
```

```python
dados = dados.apply(lambda x: x.replace('sim',True))
dados = dados.apply(lambda x: x.replace('nao',False))
```

<h1>Tratando algumas colunas e mudando tipos</h1>

<h3>Tirando as linhas com escpaço vazio em churn, visto que é a informação mais importante</h3>

```python
dados.query('Churn == ""')
```

```python
dados = dados.loc[dados['Churn']!= ""].copy()
```

<h3>Subistituindo feminino por F e masculino por M, para melhor visualização</h3>

```python
dados['cliente.genero'] = dados['cliente.genero'].apply(lambda x: x.replace('feminino','F').replace('masculino','M'))
dados
```

<h3>Todas as coluns boleanas estão como True ou False, exceto a coluna cliente.idoso, para fins de padronização vamos alterar</h3>

```python
dados['cliente.idoso'] = dados['cliente.idoso'].astype('str')
dados['cliente.idoso'] = dados['cliente.idoso'].apply(lambda x: x.replace('0','False').replace('1','True'))
dados
```

<h3>Várias colunas, como por exemplo backup online, possuem True ou False para indicar se o cliente possui esse serviço, mas também existe "sem servico de internet"</h3>
<P>Este sem serviço poderia ser também False</P>

```python
dados[['internet.backup_online','internet.protecao_dispositivo','internet.suporte_tecnico',	'internet.tv_streaming','internet.filmes_streaming']] = dados[['internet.backup_online','internet.protecao_dispositivo','internet.suporte_tecnico',	'internet.tv_streaming','internet.filmes_streaming']].astype('str')
dados[['internet.backup_online','internet.protecao_dispositivo','internet.suporte_tecnico',	'internet.tv_streaming','internet.filmes_streaming']] = dados[['internet.backup_online','internet.protecao_dispositivo','internet.suporte_tecnico',	'internet.tv_streaming','internet.filmes_streaming']].map(lambda x: x.replace('sem servico de internet','False'))
```

```python
dados.info()
```

```python
idx = dados['conta.faturamente_eletronico'].loc[(dados['conta.faturamente_eletronico']) == True & (dados['internet.servico_internet'] == False)].index
```

<h3>Outliers</h3>

```python
import seaborn as sns

sns.boxplot(x =dados['cliente.tempo_servico'])
```

```python
Q1= dados['cliente.tempo_servico'].quantile(.25)
Q3= dados['cliente.tempo_servico'].quantile(.75)
IQR = Q3-Q1
limite_inferior = Q1 - 1.5 * IQR 
limite_superior = Q3 + 1.5 * IQR 
```

```python
remover = ((dados['cliente.tempo_servico']<limite_inferior) | (dados['cliente.tempo_servico']>limite_superior))
dados = dados[~remover]
dados.reset_index(drop=True,inplace=True)
```
